```python
import pandas as pd
import numpy as np
```

## Data

```python
participants = pd.read_csv("dataset/participants.tsv", sep="\t")
participants["participant_id"] = participants["participant_id"].str.replace("sub-", "subject_")
participants.sort_values(by="participant_id", inplace=True)
participants = participants[["participant_id", "Gender", "Age","MMSE", "Set","Group"]]
participants.head()
```

```python
psd_features = pd.read_csv("dataset/PSDfeature_log.csv")
psd_features = psd_features[1:]
psd_features = psd_features.rename(columns={"Electrode": "participant_id"})
psd_features.head()
```

```python
df = pd.merge(psd_features, participants, on="participant_id")
df.sort_values(by="participant_id", inplace=True)
df["Gender"] = df["Gender"].map({"M": 1, "F": 0})
df.head()
```

```python
Train = df[df["Set"] == "Train"]
Test = df[df["Set"] == "Test"]
```

```python
feature_cols = list(df.columns)[1:-5] # drop participant_id, Set, and Group
X_train = Train[feature_cols]
X_test = Test[feature_cols]

y_train = Train["Group"]
y_test = Test["Group"]

X_train.head()
```

```python
X_train["Fp1"]
```

### Binary Data (only Healthy and AD)

```python
binary_df = df[(df['Group'] == 'A') | (df['Group'] == 'C')]
Binary_Train = binary_df[binary_df["Set"] == "Train"]
Binary_Test = binary_df[binary_df["Set"] == "Test"]

Binary_X_train = Binary_Train[feature_cols]
Binary_X_test = Binary_Test[feature_cols]

Binary_X_train = Binary_X_train.astype(float)
Binary_X_test = Binary_X_test.astype(float)

Binary_y_train = Binary_Train["Group"]
Binary_y_test = Binary_Test["Group"]
```

```python
Binary_X_train[["T5.1","T5.10","T5.18","Fp2.1","Fp2.10","Fp2.18",]].describe()
```

## Models
#### Binary Classification between Healthy and AD

```python
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import classification_report, accuracy_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

### Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

y_pred_dt = cross_val_predict(dt_model, Binary_X_train, Binary_y_train, cv=cv)

accuracy_dt = accuracy_score(Binary_y_train, y_pred_dt)
report_dt = classification_report(Binary_y_train, y_pred_dt, target_names=["Alzheimer", "Control"])

print("Decision Tree Results:")
print(f"Train Accuracy: {accuracy_dt:.4f}")
print(f"Classification Report:\n{report_dt}")
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(Binary_X_train, Binary_y_train)

y_pred_rf = cross_val_predict(rf_model, Binary_X_train, Binary_y_train, cv=cv)

accuracy_rf = accuracy_score(Binary_y_train, y_pred_rf)
report_rf = classification_report(Binary_y_train, y_pred_rf, target_names=["Alzheimer", "Control"])

print("\nRandom Forest Results:")
print(f"Train Accuracy: {accuracy_rf}")
print(f"Classification Report:\n{report_rf}")
```

### XGBoost

```python
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
xgb_model = XGBClassifier(eval_metric="mlogloss", random_state=42)

label_encoder = LabelEncoder()
Binary_y_train_xgb = label_encoder.fit_transform(Binary_y_train)

xgb_model.fit(Binary_X_train, Binary_y_train_xgb)

y_pred_xgb = cross_val_predict(xgb_model, Binary_X_train, Binary_y_train_xgb, cv=cv)

accuracy_xgb = accuracy_score(Binary_y_train_xgb, y_pred_xgb)
report_xgb = classification_report(Binary_y_train_xgb, y_pred_xgb, target_names=["Alzheimer", "Control"])

print("\nXGBoost Results:")
print(f"Train Accuracy: {accuracy_xgb}")
print(f"Classification Report:\n{report_xgb}")
```

# Hyper-Parameter Tuning + Cross Validation

```python
from sklearn.model_selection import GridSearchCV

rf_param_grid = {
    "n_estimators": [50, 100, 150, 200, 300, 400],
    "max_depth": [3, 5, 7, 9],
    "min_samples_split": [2, 4, 6, 8],
    "min_samples_leaf": [2, 4, 6]  
}

dt_param_grid = {
    "max_depth": list(np.arange(3, 11, 1)) +  [None],
    "min_samples_split": np.arange(2, 11, 1),
    "min_samples_leaf": [1, 2, 3, 4, 5]
}

xgb_param_grid = {
    "n_estimators": [50, 100, 150, 200, 300, 400],
    "max_depth": list(np.arange(3, 11, 1)) +  [None],
    "learning_rate": [0.001, 0.005, 0.01, 0.05, 0.1, 0.15, 0.2],
    "subsample": [0.8, 1.0]
}
```

```python
def cross_val_and_tune(model, param_grid, X, y, xgboost=False):
    if xgboost:
        label_encoder = LabelEncoder()
        y = label_encoder.fit_transform(y)
        X = X.astype("float32")

    grid_search = GridSearchCV(model, param_grid, cv=5, scoring="accuracy", n_jobs=-1, verbose=1)
    grid_search.fit(X, y)
    best_model = grid_search.best_estimator_
    best_params = grid_search.best_params_
    best_score = grid_search.best_score_
    return best_model, best_params, best_score
```

```python
best_rf, rf_params, rf_score = cross_val_and_tune(RandomForestClassifier(random_state=42), rf_param_grid, Binary_X_train, Binary_y_train)
best_dt, dt_params, dt_score = cross_val_and_tune(DecisionTreeClassifier(random_state=42), dt_param_grid, Binary_X_train, Binary_y_train)
best_xgb, xgb_params, xgb_score = cross_val_and_tune(XGBClassifier(eval_metric="mlogloss", random_state=42), xgb_param_grid, Binary_X_train, Binary_y_train, xgboost=True)


tuning_results = {
    "Decision Tree": {"Best Params": dt_params, "Best Score": dt_score},
    "Random Forest": {"Best Params": rf_params, "Best Score": rf_score},
    "XGBoost": {"Best Params": xgb_params, "Best Score": xgb_score}
}

tuning_df = pd.DataFrame(tuning_results).T
```
