```python
import pandas as pd
```

## Data

```python
participants = pd.read_csv("dataset/participants.tsv", sep="\t")
participants["participant_id"] = participants["participant_id"].str.replace("sub-", "subject_")
participants.sort_values(by="participant_id", inplace=True)
participants = participants[["participant_id", "Gender", "Age","MMSE", "Set","Group"]]
participants
```

```python
psd_features = pd.read_csv("dataset/PSDfeature.csv")
psd_features = psd_features[1:]
psd_features = psd_features.rename(columns={"Electrode": "participant_id"})
psd_features
```

```python
df = pd.merge(psd_features, participants, on="participant_id")
df.sort_values(by="participant_id", inplace=True)
df["Gender"] = df["Gender"].map({"M": 1, "F": 0})
df
```

```python
Train = df[df["Set"] == "Train"]
Test = df[df["Set"] == "Test"]
```

```python
feature_cols = list(df.columns)[1:-2] # drop participant_id, Set, and Group
X_train = Train[feature_cols]
X_test = Test[feature_cols]

y_train = Train["Group"]
y_test = Test["Group"]
```

### Binary Data (only Healthy and AD)

```python
binary_df = df[(df['Group'] == 'A') | (df['Group'] == 'F')]
Binary_Train = binary_df[binary_df["Set"] == "Train"]
Binary_Test = binary_df[binary_df["Set"] == "Test"]
Binary_X_train = Binary_Train[feature_cols]
Binary_X_test = Binary_Test[feature_cols]

Binary_y_train = Binary_Train["Group"]
Binary_y_test = Binary_Test["Group"]
```

```python
Binary_X_train.shape
```

## Models
#### Binary Classification between Healthy and AD

### Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

def decision_tree_classifier(X_train, y_train, X_test, y_test):
    dt_model = DecisionTreeClassifier(random_state=42)
    # dt_model = DecisionTreeClassifier()
    dt_model.fit(X_train, y_train)


    y_pred_dt = dt_model.predict(X_test)


    accuracy_dt = accuracy_score(y_test, y_pred_dt)
    # report_dt = classification_report(y_test, y_pred_dt, target_names=["Alzheimer's", "FTD", "Healthy"])
    report_dt = classification_report(y_test, y_pred_dt, target_names=["Alzheimer's", "Healthy"])

    print("Decision Tree Results:")
    print(f"Accuracy: {accuracy_dt}")
    print(f"Classification Report:\n{report_dt}")
```

```python
decision_tree_classifier(Binary_X_train, Binary_y_train, Binary_X_test, Binary_y_test)
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

def random_forest_classifier(X_train, y_train, X_test, y_test):
    rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
    rf_model.fit(X_train, y_train)


    y_pred_rf = rf_model.predict(X_test)


    accuracy_rf = accuracy_score(y_test, y_pred_rf)
    # report_rf = classification_report(y_test, y_pred_rf, target_names=["Alzheimer's", "FTD", "Healthy"])
    report_rf = classification_report(y_test, y_pred_rf, target_names=["Alzheimer's", "Healthy"])


    print("\nRandom Forest Results:")
    print(f"Accuracy: {accuracy_rf}")
    print(f"Classification Report:\n{report_rf}")
```

```python
random_forest_classifier(Binary_X_train, Binary_y_train, Binary_X_test, Binary_y_test)
```

### XGBoost

```python
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
def xgboost_classifier(X_train, y_train, X_test, y_test):
    label_encoder = LabelEncoder()

    y_train = label_encoder.fit_transform(y_train)
    y_test = label_encoder.transform(y_test)
    X_train = X_train.astype("float32")
    X_test = X_test.astype("float32")

    xgb_model = XGBClassifier(use_label_encoder=False, eval_metric="mlogloss", random_state=42)
    xgb_model.fit(X_train, y_train)

    y_pred_xgb = xgb_model.predict(X_test)

    accuracy_xgb = accuracy_score(y_test, y_pred_xgb)
    # report_xgb = classification_report(y_test, y_pred_xgb, target_names=["Alzheimer's", "FTD", "Healthy"])
    report_xgb = classification_report(y_test, y_pred_xgb, target_names=["Alzheimer's", "Healthy"])

    print("\nXGBoost Results:")
    print(f"Accuracy: {accuracy_xgb}")
    print(f"Classification Report:\n{report_xgb}")
```

```python
xgboost_classifier(Binary_X_train, Binary_y_train, Binary_X_test, Binary_y_test)
```
